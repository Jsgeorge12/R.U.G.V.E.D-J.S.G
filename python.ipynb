```python
match_runs = df1.groupby("match_id")["total_runs"].sum().reset_index()
print(match_runs)

merged_df = df2.merge(match_runs, left_on="id", right_on="match_id")

season_runs = merged_df.groupby("season")["total_runs"].sum().reset_index()

print(season_runs)
```

```python
toss_win=df2["toss_winner"].value_counts()
print(toss_win)
```

```python
toss_desc= df2.groupby('season')["toss_decision"].value_counts()
print(toss_desc)

import matplotlib.pyplot as plt

toss_desc_unstacked = toss_desc.unstack()

toss_desc_unstacked.plot(kind="bar", stacked=True, figsize=(12,6), colormap="viridis")

plt.xlabel("Season")
plt.ylabel("Count of Toss Decisions")
plt.title("Toss Decisions per Season")
plt.legend(title="Toss Decision")
plt.xticks(rotation=45)

plt.show()
```

```python
dis_kind= df1['dismissal_kind'].value_counts()
print(dis_kind)

import matplotlib.pyplot as plt
x= np.arrange
plt.figure(figsize=(8, 8))
dis_kind.plot(kind="pie", autopct="%1.1f%%", startangle=140, colormap="viridis")
plt.title("Distribution of Dismissal Kinds")
plt.ylabel("")
plt.show()

a,b = np.polyfit(dis_kind.index, dis_kind.values, 1)
plt.scatter(dis_kind.index, dis_kind.values)
plt.plot(dis_kind.index, a*dis_kind.index + b)
plt.show()
```

```python
top_ten= df1.groupby('batsman')['total_runs'].sum().reset_index().sort_values(by='total_runs', ascending=False).head(10)
print(top_ten)

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(top_ten["batsman"], top_ten["total_runs"], marker="o", linestyle="-", color="b")

plt.xlabel("Batsman")
plt.ylabel("Total Runs")
plt.title("Top 10 Batsmen by Total Runs")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
mom= df2['player_of_match'].value_counts().sort_values(ascending=False).head(10)
print(mom)

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(mom.index, mom.values, marker="o", linestyle="-", color="b")

plt.xlabel("Player")
plt.ylabel("Number of Awards")
plt.title("Top 10 Players with Most Player of the Match Awards")
plt.xticks(rotation=45)

plt.grid(True)
plt.show()
```

```python
team_1= df2['team1'].value_counts()
team_2= df2['team2'].value_counts()
total_played= team_1+team_2
print(total_played)
```

```python
winning_team= df2['winner'].value_counts()
print(winning_team)
win_rate= (winning_team/total_played)*100
print(win_rate)

import matplotlib.pyplot as plt

teams_sorted = total_played.sort_values(ascending=False).index

sorted_total_played = total_played[teams_sorted]
sorted_wins = winning_team[teams_sorted]
sorted_win_rate = win_rate[teams_sorted]

plt.figure(figsize=(12, 6))

plt.plot(teams_sorted, sorted_total_played, marker="o", linestyle="-", label="Total Matches", color="blue")

plt.plot(teams_sorted, sorted_wins, marker="s", linestyle="-", label="Wins", color="green")

plt.plot(teams_sorted, sorted_win_rate, marker="^", linestyle="-", label="Win Rate (%)", color="red")

plt.xlabel("Teams")
plt.ylabel("Count / Percentage")
plt.title("Total Matches vs Wins vs Win Rate")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)

plt.show()
```
