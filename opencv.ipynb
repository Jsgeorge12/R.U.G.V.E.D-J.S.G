```python
import cv2
import numpy as np

cap = cv2.VideoCapture(r"C:\Users\user\Downloads\Ball_Tracking.mp4")

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

output_file = 'Processed_Video.avi'
fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter(output_file, fourcc, fps, (frame_width, frame_height))

lower_green = np.array([30, 150, 50])  
upper_green = np.array([80, 255, 255]) 

while True:
    ret, frame = cap.read()
    if not ret:
        break 

    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    mask = cv2.inRange(hsv, lower_green, upper_green)

    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, np.ones((5,5), np.uint8))
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, np.ones((5,5), np.uint8))

    contours, _ = cv2.findContours(mask, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

    if contours:
        largest_contour = max(contours, key=cv2.contourArea)
        (x, y), radius = cv2.minEnclosingCircle(largest_contour)

        if radius > 10:
            cv2.circle(frame, (int(x), int(y)), int(radius), (0, 255, 0), 3)

    out.write(frame)

cap.release()
out.release()

import shutil

output_file = "Processed_Video.avi"  
destination = r"C:\Users\user\Downloads\Processed_Video.avi"  

shutil.move(output_file, destination)
print("File moved successfully!")
```
